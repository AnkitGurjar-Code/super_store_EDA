### Variables Description

**Order Information**

Row ID : Unique identifier for each record/transaction row.

Order ID : Unique identifier assigned to a customer order.

Order Date : Date on which the order was placed.

Ship Date : Date on which the order was shipped.

Ship Mode : Shipping method used for delivery (e.g., Standard Class, Second Class, First Class, Same Day).

**Customer Information**

Customer ID : Unique identifier assigned to a customer.

Customer Name : Name of the customer who placed the order.

Segment : Customer category or market segment (Consumer, Corporate, Home Office).

**Location Information**

Country : Country where the order was placed.

City : City associated with the order.

State : State associated with the order.

Postal Code : Postal/ZIP code of the customer's location.

Region : Geographic sales region (East, West, Central, South).

**Product Information**

Product ID : Unique identifier assigned to a product.

Category : High-level product category (Furniture, Office Supplies, Technology).

Sub-Category : More specific classification of products within a category.

Product Name : Name or description of the product sold.

**Sales & Performance Metrics**

Sales : Revenue generated from the sale of the product.

Quantity : Number of units sold in the transaction.

Discount : Discount applied to the product (represented as a proportion).

Profit : Profit or loss earned from the transaction after accounting for costs and discounts.

## 4. ***Questions Related to Dataset***

####which category is generating most revenue

````python
salas_cat = sales.groupby('Category')['Sales'].sum()
salas_cat
````

####Which product is selling the most?

````python
_most_sale_sub_cat = sales['Sub-Category'].value_counts().head(5)
print("The top selling sub_categories are :- \n", _most_sale_sub_cat)
````

````python
px.histogram(sales, 'Sub-Category')
````

####Which city generated the most revenue?

````python
# Assuming your DataFrame is named 'superstore_sales'
Most_sales_city = sales.groupby('City')['Sales'].sum().reset_index().sort_values(by='Sales', ascending=False).head(1)
print("The City Generates the most revenue is :- \n", Most_sales_city)
````

####profit VS sales analysis

````python
sales_sorted = sales.sort_values(by='Sales', ascending=True)
pVs = px.line(sales_sorted, x='Sales', y='Profit')
pVs.show()
````

####Discount VS Profit

````python
_fig = px.line(sales, 'Profit', 'Discount')
_fig.show()
````

####sales analysis by month

````python
# sns.barplot(data=sales, x= 'year_month', y= 'Sales')
# plt.title("Sales Analysis by Month")
# plt.xlabel("Order month")
# plt.ylabel("Sales")
# plt.show()
sales['year_month'] = sales['year_month'].astype(str)
_fig = px.bar(sales, x= 'year_month', y='Sales', title='Sales Analysis by Month')
_fig.show()
````

####Profit by Month

````python
monthly_data = sales.groupby('year_month')['Profit'].sum().reset_index()
````

````python
PbM = px.bar(monthly_data, 'Profit')
PbM.show()
````

####Year-wise Monthly Profit Trend

````python
monthly_data['year_month'] = monthly_data['year_month'].astype(str)

fig = px.line(monthly_data, x='year_month', y='Profit', title='Year-wise Monthly Profit Trend')
fig.show()
````
